```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ضبط مسار العمل ليكون في المجلد الرئيسي للمشروع بغض النظر عن مكان النوتبوك
if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')

# التأكد من وجود مجلد visuals في الجذر الرئيسي
os.makedirs("visuals", exist_ok=True)
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'sans-serif'

# تحديد مسار ملف الإكسل المعتمد
filepath = "data/processed/Saudi_Tourism_Analytics_Cleaned.xlsx"
if not os.path.exists(filepath):
    # محاولة مسار احتياطي لو كان المسار مختلفاً
    filepath = "../data/processed/Saudi_Tourism_Analytics_Cleaned.xlsx"
    if not os.path.exists(filepath):
        raise FileNotFoundError("Error: The processed Excel dataset could not be located in 'data/processed/'.")

print("Info: Reading dataset and generating all 4 core project visualizations...")

# 1. Top Regions by Total Available Rooms
try:
    df_reg = pd.read_excel(filepath, sheet_name="Regional_KPIs")
    top_regions = df_reg.sort_values(by="2_2_Total_available_rooms", ascending=False).head(5)

    plt.figure(figsize=(10, 5))
    sns.barplot(data=top_regions, x="2_2_Total_available_rooms", y="Region", palette="Blues_r")
    plt.title("Top 5 Regions by Total Available Hotel Rooms (Q2 2026)", fontsize=14, fontweight='bold')
    plt.xlabel("Total Available Rooms")
    plt.ylabel("Region")
    plt.tight_layout()
    plt.savefig("visuals/top_regions_rooms.png", dpi=300)
    plt.close()
    print("Success: Generated and saved 'top_regions_rooms.png'.")
except Exception as e:
    print(f"Warning: Failed to process top regions plot. Details: {e}")

# 2. Total Employment Distribution Across Tourism Activities
try:
    df_emp = pd.read_excel(filepath, sheet_name="Employment_Activity")

    plt.figure(figsize=(12, 6))
    sns.barplot(data=df_emp.sort_values(by="Total_Total", ascending=False), x="Total_Total", y="Activity", palette="viridis")
    plt.title("Total Employment Distribution Across Tourism Activities (Q2 2026)", fontsize=14, fontweight='bold')
    plt.xlabel("Total Employees (Saudi & Non-Saudi)")
    plt.ylabel("Tourism Activity")
    plt.tight_layout()
    plt.savefig("visuals/employment_by_activity.png", dpi=300)
    plt.close()
    print("Success: Generated and saved 'employment_by_activity.png'.")
except Exception as e:
    print(f"Warning: Failed to process employment distribution plot. Details: {e}")

# 3. Quarterly Hotel Occupancy Rate Trends
try:
    df_trends = pd.read_excel(filepath, sheet_name="Quarterly_Trends")
    
    if "Occupancy_Rate" in df_trends.columns and "Quarter" in df_trends.columns:
        plt.figure(figsize=(12, 5))
        sns.lineplot(data=df_trends, x="Quarter", y="Occupancy_Rate", marker="o", color="#1f77b4", linewidth=2.5)
        plt.title("Hotel Occupancy Rate Trends Over Quarters", fontsize=14, fontweight='bold')
        plt.xlabel("Quarter")
        plt.ylabel("Occupancy Rate (%)")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.savefig("visuals/occupancy_rate_trends.png", dpi=300)
        plt.close()
        print("Success: Generated and saved 'occupancy_rate_trends.png'.")
except Exception as e:
    print(f"Warning: Failed to process occupancy trends plot. Details: {e}")

# 4. Saudization Rate Analysis Across Tourism Activities
try:
    df_emp_saudi = pd.read_excel(filepath, sheet_name="Employment_Activity")
    
    if "Saudi_Total" in df_emp_saudi.columns and "Total_Total" in df_emp_saudi.columns:
        df_emp_saudi["Saudization_Rate"] = (df_emp_saudi["Saudi_Total"] / df_emp_saudi["Total_Total"]) * 100
        df_emp_saudi = df_emp_saudi.sort_values(by="Saudization_Rate", ascending=False)

        plt.figure(figsize=(12, 6))
        sns.barplot(data=df_emp_saudi, x="Saudization_Rate", y="Activity", palette="crest")
        plt.title("Saudization Rate (%) Across Tourism Activities", fontsize=14, fontweight='bold')
        plt.xlabel("Saudization Rate (%)")
        plt.ylabel("Tourism Activity")
        plt.tight_layout()
        plt.savefig("visuals/saudization_by_activity.png", dpi=300)
        plt.close()
        print("Success: Generated and saved 'saudization_by_activity.png'.")
except Exception as e:
    print(f"Warning: Failed to process Saudization rate plot. Details: {e}")

print("Status: All 4 visualization tasks completed successfully. Output files stored in the main 'visuals' directory.")
```
